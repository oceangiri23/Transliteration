```python
def merge_txt_files(file_list, output_file):
    with open(output_file, 'w', encoding='utf-8') as outfile:
        for file_name in file_list:
            with open(file_name, 'r', encoding='utf-8') as infile:
                outfile.write(infile.read())
                outfile.write("\n")  
            print(f"Merged: {file_name}")
    print(f"\nAll files merged into '{output_file}' successfully.")

files = [
    "GadgetByteNepali_comments_romanized.txt",
    "merged_comments_highlightsNepal.txt",
    "TechLekhYT_comments_romanized.txt",
    "../Normalization/data_text_files/clean_corpus.txt",
    "../Normalization/data_text_files/merged.txt",
]

output = "all_merged.txt"

merge_txt_files(files, output)
```

```python


# count_by_line.py

def count_lines_in_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        # Read all non-empty lines
        lines = [line.strip() for line in f if line.strip()]
    return len(lines), lines



# Example for multiple files
files = ["all_merged.txt"]
total = 0
for f in files:
    c, _ = count_lines_in_file(f)
    total += c
    print(f"{f}: {c} lines")

print(f"\nTotal lines across all files: {total}")
```
