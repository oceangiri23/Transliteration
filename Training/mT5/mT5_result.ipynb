```python
import matplotlib.pyplot as plt

# Data
epochs = list(range(10))

train_loss = [2.281, 0.729, 0.402, 0.232, 0.138, 0.085, 0.055, 0.037, 0.027, 0.019]
val_loss   = [1.0150, 0.5509, 0.3526, 0.2410, 0.1910, 0.1335, 0.1100, 0.0895, 0.0785, 0.0700]

cer = [0.2845, 0.1548, 0.1039, 0.0840, 0.0697, 0.0486, 0.0418, 0.0306, 0.0222, 0.0243]

word_acc = [0.4455, 0.6178, 0.7183, 0.7580, 0.8135, 0.8561, 0.8796, 0.9096, 0.9327, 0.9343]

bleu5 = [0.6702, 0.8101, 0.8658, 0.8928, 0.9065, 0.9341, 0.9448, 0.9607, 0.9683, 0.9674]


# Smaller marker size
ms = 3


# 1. Train vs Validation Loss
plt.figure()
plt.plot(epochs, train_loss, marker='o', markersize=ms, label='Train Loss')
plt.plot(epochs, val_loss, marker='o', markersize=ms, label='Validation Loss')

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Train vs Validation Loss")

plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()


# 2. CER vs Epoch
plt.figure()
plt.plot(epochs, cer, marker='o', markersize=ms)

plt.xlabel("Epoch")
plt.ylabel("CER")
plt.title("CER vs Epoch")

plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


# 3. Word Accuracy vs Epoch
plt.figure()
plt.plot(epochs, word_acc, marker='o', markersize=ms)

plt.xlabel("Epoch")
plt.ylabel("Word Accuracy")
plt.title("Word Accuracy vs Epoch")

plt.grid(True, linestyle='--', alpha=0.6)
plt.show()


# 4. Average BLEU-5 vs Epoch
plt.figure()
plt.plot(epochs, bleu5, marker='o', markersize=ms)

plt.xlabel("Epoch")
plt.ylabel("Average BLEU-4")
plt.title("Average BLEU-4 vs Epoch")

plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
