```python
import pandas as pd
import random

def sample_csv_columns(file_path, n_samples=200):
    """
    Randomly sample n sentences from the first and second columns of a CSV file
    
    Args:
        file_path (str): Path to the CSV file
        n_samples (int): Number of samples to take (default 200)
    
    Returns:
        tuple: Two lists containing samples from first and second columns
    """
    # Read the CSV file
    df = pd.read_csv(file_path)
    
    # Get the total number of rows
    total_rows = len(df)
    
    # Determine how many samples we can actually take
    actual_samples = min(n_samples, total_rows)
    
    # Generate random indices
    random_indices = random.sample(range(total_rows), actual_samples)
    
    # Extract values from first and second columns
    first_column_list = [df.iloc[i, 0] for i in random_indices]
    second_column_list = [df.iloc[i, 1] for i in random_indices]
    
    return first_column_list, second_column_list

# Example usage
if __name__ == "__main__":
    # Replace 'your_file.csv' with your actual CSV file path
    file_path = 'C:\\Users\\N I T R O V15\\OneDrive\\Documents\\Major Project\\Transliteration\\Final Datasets\\NEW Parallel\\final_parallel_dedup.csv'
    
    try:
        first_col_samples, second_col_samples = sample_csv_columns(file_path, 200)
        
        print("First Column Samples:")
        print(first_col_samples)
        print("\nSecond Column Samples:")
        print(second_col_samples)
        
    except FileNotFoundError:
        print(f"File {file_path} not found. Please provide the correct path.")
    except IndexError:
        print("The CSV file doesn't have enough columns or rows.")
    except Exception as e:
        print(f"An error occurred: {e}")
```
