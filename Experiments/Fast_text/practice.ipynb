```python
import fasttext
import re
import os

model = fasttext.load_model("/capstor/store/cscs/swissai/a168/sagargiri/Fast_text/fasttext_devanagari.bin")

def similar_words(word, k=5):
    neighbors = model.get_nearest_neighbors(word, k=k)
    print(f"\nTop {k} words similar to '{word}':")
    for score, w in neighbors:
        print(f"  {w:20s}  →  {score:.4f}")
```

```python
similar_words("गायेरा")
```
