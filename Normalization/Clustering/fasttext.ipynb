```python
from gensim.models import FastText
import os
import numpy as np
```

```python
corpus_path = "../data_text_files/clean_corpus.txt"
model_path =  "Fast_text_model_for_clean_corpus/fasttext.model"


with open(corpus_path, 'r', encoding='utf-8') as f:
    sentences = [line.strip().split() for line in f if line.strip()]


model = FastText(
    sentences,
    vector_size=150,
    window=5,
    min_count=2,
    epochs=10,
    sg=1,  
    workers=os.cpu_count() or 1  
)

model.save(model_path)
```

```python

model_path ="Fast_text_model_for_clean_corpus/fasttext.model"

try:
    loaded_model = FastText.load(model_path)
    print("Model loaded successfully!")

except Exception as e:
    print(f"Error loading the model: {e}")
    print(f"Please ensure the model file exists at: {model_path}")
```

```python

word = "socham"
try:
    word_vector = model.wv.get_vector(word)
    print(f"\nVector for '{word}':")
    print(word_vector[:10])

    #Find the most similar words to a word
    similar_words = model.wv.most_similar(word)
    print(f"\nWords most similar to '{word}':")
    for sim_word, score in similar_words: # Note the order of results in gensim's most_similar
        print(f"{sim_word} (similarity: {score:.4f})")

except KeyError as e: # Use KeyError for missing words in gensim
    print(f"Error: {e}. The word '{word}' might not be in the model's vocabulary.")
```
