```python
import re

def text_stats(file_path):
    # Read the text file
    with open(file_path, 'r', encoding='utf-8') as f:
        text = f.read()

    # Count number of lines (non-empty lines)
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f if line.strip()]
    num_lines = len(lines)

    # Split into sentences using regex
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    # Split into words
    words = re.findall(r'\b\w+\b', text.lower())

    # Count stats
    num_sentences = len(sentences)
    num_words = len(words)
    num_unique_words = len(set(words))

    # Print results
    print(f"Number of lines: {num_lines}")
    print(f"Number of sentences: {num_sentences}")
    print(f"Number of words: {num_words}")
    print(f"Number of unique words: {num_unique_words}")

# Example usage:
text_stats("all_merged.txt")

```

```python
import json

def count_json_elements(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        data = json.load(f)

    # Handle different JSON structures
    if isinstance(data, list):
        count = len(data)
    elif isinstance(data, dict):
        count = len(data.keys())
    else:
        count = 1  # for single values like string, int, etc.

    print(f"Number of elements in JSON file: {count}")

# Example usage:
count_json_elements("../Clustering/Fresh_Complete_phonetic_Strcuctured_group.json")
```
