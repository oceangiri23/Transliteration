```python
import json

input_file = "Complete_phonetic_Dictionary.json"

range1 = (0, 6500)
range2 = (6500, 13000)

with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

items = list(data.items())

def extract_range(items, start, end):
    return dict(items[start:end])

part1 = extract_range(items, range1[0], range1[1])
part2 = extract_range(items, range2[0], range2[1])
part3 = dict(items[range2[1]:])  

with open("Saksham.json", "w", encoding="utf-8") as f:
    json.dump(part1, f, ensure_ascii=False, indent=2)

with open("Sagar.json", "w", encoding="utf-8") as f:
    json.dump(part2, f, ensure_ascii=False, indent=2)

with open("Nabin.json", "w", encoding="utf-8") as f:
    json.dump(part3, f, ensure_ascii=False, indent=2)

print("Splitting completed!")
```
